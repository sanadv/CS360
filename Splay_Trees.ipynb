```python



class Node:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None
        self.parent = None


class SplayTree:
    def __init__(self, trace=False):
        self.root = None
        self.trace = trace  # print each splay step when True

    # ---------- rotations ----------

    def _rotate_left(self, x):
        """x's right child y moves up; x becomes y's left child."""
        y = x.right
        x.right = y.left
        if y.left:
            y.left.parent = x
        y.parent = x.parent
        if x.parent is None:
            self.root = y
        elif x is x.parent.left:
            x.parent.left = y
        else:
            x.parent.right = y
        y.left = x
        x.parent = y

    def _rotate_right(self, x):
        """x's left child y moves up; x becomes y's right child."""
        y = x.left
        x.left = y.right
        if y.right:
            y.right.parent = x
        y.parent = x.parent
        if x.parent is None:
            self.root = y
        elif x is x.parent.right:
            x.parent.right = y
        else:
            x.parent.left = y
        y.right = x
        x.parent = y

    # ---------- splay ----------

    def _splay(self, x):
        while x.parent is not None:
            p = x.parent
            g = p.parent

            if g is None:
                # ZIG: parent is the root -> single rotation
                self._log(f"zig at {x.key}")
                if x is p.left:
                    self._rotate_right(p)
                else:
                    self._rotate_left(p)

            elif x is p.left and p is g.left:
                # ZIG-ZIG (left-left): rotate grandparent first, then parent
                self._log(f"zig-zig (left-left) at {x.key}")
                self._rotate_right(g)
                self._rotate_right(p)

            elif x is p.right and p is g.right:
                # ZIG-ZIG (right-right)
                self._log(f"zig-zig (right-right) at {x.key}")
                self._rotate_left(g)
                self._rotate_left(p)

            elif x is p.right and p is g.left:
                # ZIG-ZAG (left-right): rotate parent, then grandparent
                self._log(f"zig-zag (left-right) at {x.key}")
                self._rotate_left(p)
                self._rotate_right(g)

            else:
                # ZIG-ZAG (right-left)
                self._log(f"zig-zag (right-left) at {x.key}")
                self._rotate_right(p)
                self._rotate_left(g)

    # ---------- insertion ----------

    def insert(self, key):
        # Phase 1: standard BST insert
        parent = None
        cur = self.root
        while cur is not None:
            parent = cur
            if key < cur.key:
                cur = cur.left
            elif key > cur.key:
                cur = cur.right
            else:
                # Duplicate: splay the existing node and stop
                self._log(f"{key} already present; splaying it")
                self._splay(cur)
                return

        node = Node(key)
        node.parent = parent
        if parent is None:
            self.root = node
        elif key < parent.key:
            parent.left = node
        else:
            parent.right = node

        # Phase 2: splay the new node to the root
        self._splay(node)

    # ---------- helpers for display ----------

    def _log(self, msg):
        if self.trace:
            print("   ", msg)

    def inorder(self):
        out = []

        def walk(n):
            if n:
                walk(n.left)
                out.append(n.key)
                walk(n.right)

        walk(self.root)
        return out

    def show(self):
        """Print the tree sideways: root on the left, right subtree on top."""
        def walk(n, prefix, is_right):
            if n is None:
                return
            walk(n.right, prefix + ("    " if is_right else "│   "), True)
            print(prefix + ("┌── " if is_right else "└── ") + str(n.key))
            walk(n.left, prefix + ("│   " if is_right else "    "), False)

        if self.root is None:
            print("(empty)")
            return
        walk(self.root.right, "", True)
        print(str(self.root.key))
        walk(self.root.left, "", False)


if __name__ == "__main__":
    tree = SplayTree(trace=True)
    for k in [50, 30, 70, 20, 40, 60, 80, 35]:
        print(f"\nInsert {k}:")
        tree.insert(k)
        tree.show()
        assert tree.root.key == k, "inserted key must end up at the root"

    print("\nIn-order:", tree.inorder())
```

```python
class Node:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None
        self.parent = None


class SplayTree:
    def __init__(self, trace=False):
        self.root = None
        self.trace = trace  # print each splay step when True

    # ---------- rotations ----------

    def _rotate_left(self, x):
        """x's right child y moves up; x becomes y's left child."""
        y = x.right
        x.right = y.left
        if y.left:
            y.left.parent = x
        y.parent = x.parent
        if x.parent is None:
            self.root = y
        elif x is x.parent.left:
            x.parent.left = y
        else:
            x.parent.right = y
        y.left = x
        x.parent = y

    def _rotate_right(self, x):
        """x's left child y moves up; x becomes y's right child."""
        y = x.left
        x.left = y.right
        if y.right:
            y.right.parent = x
        y.parent = x.parent
        if x.parent is None:
            self.root = y
        elif x is x.parent.right:
            x.parent.right = y
        else:
            x.parent.left = y
        y.right = x
        x.parent = y

    # ---------- splay ----------

    def _splay(self, x):
        while x.parent is not None:
            p = x.parent
            g = p.parent

            if g is None:
                # ZIG: parent is the root -> single rotation
                self._log(f"zig at {x.key}")
                if x is p.left:
                    self._rotate_right(p)
                else:
                    self._rotate_left(p)

            elif x is p.left and p is g.left:
                # ZIG-ZIG (left-left): rotate grandparent first, then parent
                self._log(f"zig-zig (left-left) at {x.key}")
                self._rotate_right(g)
                self._rotate_right(p)

            elif x is p.right and p is g.right:
                # ZIG-ZIG (right-right)
                self._log(f"zig-zig (right-right) at {x.key}")
                self._rotate_left(g)
                self._rotate_left(p)

            elif x is p.right and p is g.left:
                # ZIG-ZAG (left-right): rotate parent, then grandparent
                self._log(f"zig-zag (left-right) at {x.key}")
                self._rotate_left(p)
                self._rotate_right(g)

            else:
                # ZIG-ZAG (right-left)
                self._log(f"zig-zag (right-left) at {x.key}")
                self._rotate_right(p)
                self._rotate_left(g)

    # ---------- insertion ----------

    def insert(self, key):
        # Phase 1: standard BST insert
        parent = None
        cur = self.root
        while cur is not None:
            parent = cur
            if key < cur.key:
                cur = cur.left
            elif key > cur.key:
                cur = cur.right
            else:
                # Duplicate: splay the existing node and stop
                self._log(f"{key} already present; splaying it")
                self._splay(cur)
                return

        node = Node(key)
        node.parent = parent
        if parent is None:
            self.root = node
        elif key < parent.key:
            parent.left = node
        else:
            parent.right = node

        # Phase 2: splay the new node to the root
        self._splay(node)

    # ---------- search ----------

    def search(self, key):
        """Return True if key is in the tree.

        Found: the node is splayed to the root.
        Not found: the last node visited is splayed instead, so a
        repeated failed search also becomes cheap.
        """
        cur = self.root
        last = None
        while cur is not None:
            last = cur
            if key < cur.key:
                cur = cur.left
            elif key > cur.key:
                cur = cur.right
            else:
                self._log(f"found {key}; splaying it")
                self._splay(cur)
                return True

        if last is not None:
            self._log(f"{key} not found; splaying last visited node {last.key}")
            self._splay(last)
        return False

    # ---------- deletion ----------

    def delete(self, key):
        """Remove key from the tree. Return True if it was there.

        1. Search for key, which splays it to the root.
        2. Remove the root, leaving two subtrees L (all keys < key)
           and R (all keys > key).
        3. Join: splay the largest key in L to the top of L. It now has
           no right child, so R is attached as its right subtree.
        """
        # Phase 1: bring key to the root (or stop if it is missing)
        if not self.search(key):
            self._log(f"{key} not in tree; nothing to delete")
            return False

        # Phase 2: detach the two subtrees of the root
        old = self.root
        L, R = old.left, old.right
        if L:
            L.parent = None
        if R:
            R.parent = None
        old.left = old.right = None

        # Phase 3: join L and R
        if L is None:
            self.root = R
        else:
            self.root = L
            m = L
            while m.right is not None:   # maximum of L
                m = m.right
            self._log(f"joining: splaying max of left subtree, {m.key}")
            self._splay(m)               # m becomes root of L, no right child
            m.right = R
            if R:
                R.parent = m
        return True

    # ---------- helpers for display ----------

    def _log(self, msg):
        if self.trace:
            print("   ", msg)

    def inorder(self):
        out = []

        def walk(n):
            if n:
                walk(n.left)
                out.append(n.key)
                walk(n.right)

        walk(self.root)
        return out

    def show(self):
        """Print the tree sideways: root on the left, right subtree on top."""
        def walk(n, prefix, is_right):
            if n is None:
                return
            walk(n.right, prefix + ("    " if is_right else "│   "), True)
            print(prefix + ("┌── " if is_right else "└── ") + str(n.key))
            walk(n.left, prefix + ("│   " if is_right else "    "), False)

        if self.root is None:
            print("(empty)")
            return
        walk(self.root.right, "", True)
        print(str(self.root.key))
        walk(self.root.left, "", False)


if __name__ == "__main__":
    tree = SplayTree(trace=True)
    for k in [50, 30, 70, 20, 40, 60, 80, 35]:
        print(f"\nInsert {k}:")
        tree.insert(k)
        tree.show()
        assert tree.root.key == k, "inserted key must end up at the root"

    print("\nIn-order:", tree.inorder())

    for k in [20, 65]:
        print(f"\nSearch {k}:")
        found = tree.search(k)
        print("   found" if found else "   not found")
        tree.show()

    for k in [40, 50, 99]:
        print(f"\nDelete {k}:")
        removed = tree.delete(k)
        print("   removed" if removed else "   not in tree")
        tree.show()
        assert k not in tree.inorder()

    print("\nIn-order:", tree.inorder())
```
